# The Monty Hall Problem

### In the "Monty Hall Problem," you imagine yourself on a game show where the host, Monty Hall, presents you with three doors to choose from. One door has a nice car behind it, whilst the other two have goats. You have to choose one door, but you don't know what's behind them. Your initial choice is therefore a blind guess.

### However, after you make your choice, Monty Hall will open one of the remaining doors, one that you didn't choose and that has a goat behind it. He will then offer you the opportunity to switch doors. The question is: should you switch, or should you stay with your original choice? The correct answer is that you should always switch. Explaining why this is, however, can be difficult and often frustrates the person you are trying to explain it to.

### Therefore, I will show how to simulate the game to prove that you should always switch.


```python
import numpy as np
```

First of I create a class to store the game variables and methods.

```python
# creates a class for the monty hall game
class monty_hall:
    # initiates np arrays for doors and choice
    def __init__(self):
        # vector where 1 represents the car and 0's represents the goats
        self.doors = np.array([1, 0, 0])
        # vector where 1 corresponds to the door chosen (random)
        self.choice = np.random.permutation([1, 0, 0])
        # result integer
        self.won = int()
    # method that "opens" one of the doors i.e. removing it
    def open_door(self):
        #  finda all indices where both vectors are 0 i.e. door with goat that player hasn't chosen
        both_zero_indices = np.where((self.doors == 0) & (self.choice == 0))[0]

        index_to_remove = both_zero_indices[0]  # Takes only the first match
    
        # Remove that single index from both vectors
        self.doors = np.delete(self.doors, index_to_remove)
        self.choice = np.delete(self.choice, index_to_remove)
    # method for switching doors
    def switch(self):
        self.choice = self.choice[::-1]
    # method for showing results (1 = car, 0 = goat)
    def result(self):
        self.won = self.choice @ self.doors
        return self.won
        
```

Then I create a simulation function

```python
def run_monty_hall_simulation(N):
    stay = []
    switch = []
    for n in range(N):
        # simulate 2 what if scenarios in 1 go.

        # in scenario 1 we always stay
        game = monty_hall()
        game.open_door()
        game.result()
        win_stay  = game.won
        stay.append(win_stay)
        # in scenario 2 we always switch    
        game.switch()
        game.result()
        win_switch = game.won
        switch.append(win_switch)

    print(f"stay win % : {round((sum(stay) / N) * 100, 2)}%")
    print(f"switch win % : {round((sum(switch) / N) * 100, 2)}%")


```

As the number of simulations $N$ increases, the winning percentages not only converge toward the true theoretical probabilities ($\frac{1}{3}$ for staying and $\frac{2}{3}$ for switching) but they also become much more stable, smoothing out the random fluctuations seen at smaller sample sizes.

```python
for n in [10, 100, 1000, 10000]:
    print(f"N = {n}")
    run_monty_hall_simulation(n)
```

If you run the previous cell multiple times you can see that the percentages fluctuate a lot more for smaller sample sizes.
